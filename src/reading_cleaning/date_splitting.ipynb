```python
import pandas as pd
from pathlib import Path

import os
os.chdir(r"C:\Users\jonah\Desktop\personalProjects\temp_electricity")
```

```python
df = pd.read_parquet("data/raw/fsa_temperature_hourly.parquet")
```

```python
# Loop through each month and year after 2022
df = df[df["time_utc"].dt.year > 2021]

df["hour"] = df["time_utc"].dt.hour + 1
df["date"] = df["time_utc"].dt.normalize()

df = df.drop(columns = ['time_utc', 'time_est'])
```

```python
# Output directory
output_dir = Path("data/raw")
output_dir.mkdir(parents=True, exist_ok=True)

for (year, month), monthly_df in df.groupby([
    df["date"].dt.year,
    df["date"].dt.month
]):
    # Create filename, e.g. 2023-01.parquet
    filename = output_dir / f"{year}-{month:02d}-temp.parquet"

    # Save monthly data
    monthly_df.to_parquet(filename, index=False)

    print(f"Saved {filename} ({len(monthly_df)} rows)")
```
